## Pengertian
Ada beberapa istilah yang ada di praktikum pekan satu ini:
1. print ()
=> Fungsi bawaan (built-in function) Python yang digunakan untuk menampilkan data atau teks ke layar (output standar).

2. Komentar (comment)
=> Baris kode yang tidak dieksekusi oleh interpreter, digunakan untuk memberi keterangan pada kode. Python mendukung komentar satu baris (#) dan komentar banyak baris menggunakan string tiga kutip ("""...""" atau '''...''') yang tidak diberi nama variabel.

3. string
=> Tipe data yang berupa kumpulan karakter, ditulis di antara tanda kutip tunggal ('...') atau ganda ("...").

4. Concatenation (Penggabungan String)
=> Proses menggabungkan dua string atau lebih menjadi satu string baru, pada Python dilakukan menggunakan operator

5. Escape Character
=> 	Karakter khusus yang diawali tanda backslash (\) yang memiliki fungsi/makna tertentu dalam string, misalnya \n untuk baris baru dan \\ untuk menampilkan satu karakter backslash.

```python
# ==========================================
# PROGRAM KASIR SEDERHANA
# ==========================================

print("======================================")
print("       PROGRAM KASIR SEDERHANA")
print("======================================")

# Daftar barang
barang = {
    1: ("Buku", 10000),
    2: ("Pulpen", 5000),
    3: ("Pensil", 3000),
    4: ("Penghapus", 2000),
    5: ("Penggaris", 7000)
}

total = 0

# Menampilkan daftar barang
print("\nDAFTAR BARANG")
print("--------------------------------------")

for nomor, (nama, harga) in barang.items():
    print(f"{nomor}. {nama:<15} Rp{harga:,}")

print("--------------------------------------")

# Memilih barang
while True:
    pilihan = input("\nMasukkan nomor barang (0 untuk selesai): ")

    # Mengecek apakah input berupa angka
    if not pilihan.isdigit():
        print("Masukkan nomor yang benar!")
        continue

    pilihan = int(pilihan)

    # Jika 0, selesai memilih
    if pilihan == 0:
        break

    # Mengecek nomor barang
    if pilihan not in barang:
        print("Barang tidak tersedia!")
        continue

    nama, harga = barang[pilihan]

    # Memasukkan jumlah barang
    jumlah_input = input(f"Jumlah {nama}: ")

    if not jumlah_input.isdigit():
        print("Jumlah harus berupa angka!")
        continue

    jumlah = int(jumlah_input)

    subtotal = harga * jumlah
    total += subtotal

    print(f"{nama} x {jumlah} = Rp{subtotal:,}")
    print(f"Total sementara: Rp{total:,}")

# Jika tidak ada barang yang dibeli
if total == 0:
    print("\nTidak ada barang yang dibeli.")
    print("Terima kasih!")

else:
    # Menampilkan total
    print("\n======================================")
    print(f"TOTAL BELANJA : Rp{total:,}")
    print("======================================")

    # Pembayaran
    while True:
        uang_input = input("Masukkan uang pembayaran: ")

        if not uang_input.isdigit():
            print("Masukkan jumlah uang yang benar!")
            continue

        uang = int(uang_input)

        if uang < total:
            print("Uang tidak cukup!")
            print(f"Anda masih kurang Rp{total - uang:,}")
        else:
            break

    # Menghitung kembalian
    kembalian = uang - total

    print("\n======================================")
    print("          STRUK PEMBAYARAN")
    print("======================================")
    print(f"Total Belanja : Rp{total:,}")
    print(f"Uang Bayar    : Rp{uang:,}")
    print(f"Kembalian     : Rp{kembalian:,}")
    print("======================================")
    print("     TERIMA KASIH SUDAH BELANJA!")
    print("======================================")
```
