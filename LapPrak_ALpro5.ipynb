## 2.1 Dasar Teori
Perulangan adalah struktur kontrol yang menjalankan suatu blok perintah berulang kali selama kondisi tertentu terpenuhi atau sebanyak jumlah yang ditentukan. Perulangan mengurangi pengulangan penulisan kode dan membuat program lebih ringkas.

Pada Python, perulangan for menjalankan blok perintah untuk setiap elemen dalam sebuah deret. Fungsi range() menghasilkan deret bilangan bulat dan memiliki tiga bentuk:
- range(n) menghasilkan bilangan dari 0 sampai n - 1.
- range(awal, akhir) menghasilkan bilangan dari awal sampai akhir - 1.
- range(awal, akhir, langkah) menghasilkan bilangan dari awal sampai akhir - 1 dengan kenaikan sebesar langkah.
Batas akhir pada range() tidak ikut dihitung, sehingga untuk mencapai bilangan n dari 1 digunakan range(1, n + 1).

Fungsi input() membaca masukan pengguna dan selalu mengembalikan nilai bertipe string. Fungsi int() mengubah string menjadi bilangan bulat. Fungsi print() menampilkan keluaran dan secara bawaan berpindah baris setelah mencetak. Parameter end dapat mengubah perilaku ini, misalnya end=" " mencetak spasi dan end="" tidak mencetak apa pun sehingga keluaran berikutnya berada di baris yang sama. F-stri

Variabel counter menghitung banyaknya suatu kejadian, yaitu bertambah 1 pada setiap perulangan (jumlah = jumlah + 1). Variabel akumulator mengumpulkan total nilai, yaitu menambahkan nilai variabel perulangan (jumlah = jumlah + i). Perbedaan keduanya menentukan apakah hasil akhirnya berupa banyaknya data atau jumlah nilai data.

Percabangan (if, elif, else) menjalankan blok perintah berdasarkan kondisi. Operator modulus (%) menghasilkan sisa bagi. Suatu bilangan adalah genap jika sisa baginya dengan 2 sama dengan 0 (i % 2 == 0)

Perulangan bersarang adalah perulangan yang berada di dalam perulangan lain. Setiap satu putaran perulangan luar akan menjalankan seluruh putaran perulangan dalam. Jika perulangan luar berjalan m kali dan perulangan dalam n kali, maka blok terdalam dijalankan m x n kali. Perulangan luar umumnya mewakili baris, sedangkan perulangan dalam mewakili kolom.

Catatan:
- Perbedaan kedua perulangan hanya pada titik awal: range(n) mulai dari 0, sedangkan range(1, n+1) mulai dari 1. Keduanya sama-sama berjalan sebanyak n kali.

```python
ulang_3015 = int(input("Masukkan jumlah perulangan: "))

jumlah_3015 = 0
for i_3015 in range (1, ulang_3015 + 1):
    print (i_3015, end=" ")
    jumlah_3015 = jumlah_3015 + 1

    if i_3015 < ulang_3015:
        print (" + ", end="")
    else:
        print (" = ", jumlah_3015, end="")
print ()
print ("Jumlah = ", jumlah_3015)
```

```python
ulang_3015 = int (input("Masukkan nilai batas: "))

jumlah_3015 = 0
for i_3015 in range (1, ulang_3015 + 1):
    if i_3015 % 2 == 0:
        print (i_3015, end=" ")
        jumlah_3015 = jumlah_3015 + i_3015

        if i_3015 < ulang_3015:
            print (" + ", end="")
        else:
            print (" = ", jumlah_3015, end="")
print ()
print ("Jumlah = ", jumlah_3015)
```

3.     for j_3015 in range (batas_3015 - i_3015):
   - Perulangan dalam pertama, khusus untuk mencetak spasi di sebelah kiri.
   - Berjalan sebanyak (batas_3015 - i_3015) kali, jadi makin besar nomor baris, makin sedikit putarannya.
   - Contoh: jika batas 4 dan baris ke-1, berjalan 4 - 1 = 3 kali. Pada baris terakhir (i = batas), range(0) kosong sehingga tidak ada spasi.
   - Variabel j_3015 hanya penghitung putaran, nilainya tidak dipakai.

4.         print("  ", end="")
   - Mencetak 2 spasi tanpa pindah baris.
   - Karena dijalankan (batas - i) kali, spasi di kiri berjumlah 2 x (batas - i).

5.     for k_3015 in range (i_3015):
   - Perulangan dalam kedua, khusus untuk mencetak bintang.
   - range(i_3015) menghasilkan i_3015 putaran, jadi jumlah bintang sama dengan nomor baris.
   - Variabel k_3015 hanya penghitung putaran, nilainya tidak dipakai.

6.         print(" * ", end=" ")
   - Mencetak " * " (spasi, bintang, spasi), lalu ditambah satu spasi dari end=" ".
   - Jadi setiap bintang menempati 4 karakter: " *  ". Hasilnya, antar bintang ada 3 spasi.

7.     print () # pindah ke baris berikutnya
   - Berada di dalam perulangan luar, tetapi di luar kedua perulangan dalam.
   - Dijalankan sekali setelah satu baris selesai, berfungsi sebagai pindah baris.
   - Teks setelah tanda # adalah komentar, tidak dijalankan oleh Python.

Contoh jalannya program:

Jika pengguna memasukkan angka 4, maka keluarannya:
       *  
     *   *  
   *   *   *  
 *   *   *   *  

Rincian per baris (batas = 4):
- Baris 1: 3 x 2 = 6 spasi, lalu 1 bintang
- Baris 2: 2 x 2 = 4 spasi, lalu 2 bintang
- Baris 3: 1 x 2 = 2 spasi, lalu 3 bintang
- Baris 4: 0 spasi, lalu 4 bintang

Catatan:
- Bentuknya menjadi piramida simetris karena lebar tiap bintang (4 karakter) dua kali lebih besar daripada pergeseran spasi tiap baris (2 karakter). Setiap baris turun, spasi kiri berkurang 2, sehingga bintang paling kiri bergeser 2 kolom ke kiri, sementara bintang paling kanan bergeser 2 kolom ke kanan.
- Meskipun teks input menyebut "bilangan genap", kode tidak memeriksa hal itu. Bilangan ganjil pun tetap menghasilkan piramida yang benar.
- Di akhir setiap baris ada spasi tambahan dari end=" " pada bintang terakhir. Spasi itu tidak terlihat dan tidak memengaruhi tampilan.
- Perulangan dalam dengan j_3015 dan k_3015 bisa diganti dengan perkalian string, misalnya: print("  " * (batas_3015 - i_3015) + " *  " * i_3015). Hasilnya sama.
- Jika pengguna memasukkan teks yang bukan angka, program error (ValueError).
- Jika pengguna memasukkan 0 atau bilangan negatif, perulangan luar tidak berjalan sehingga tidak ada yang tercetak.

Mahasiswa diminta merancang program konsol interaktif untuk mencetak pola bangun geometri dua dimensi berupa Jam Pasir Kristal Palindromik yang terbungkus rapi di dalam bingkai pembatas berpagar. Berbeda dengan pola dasar, program ini menuntut pembentukan deret angka yang bergerak dinamis menurun dan menaik secara palindromik pada setiap baris serta diselaraskan dengan poros kristal di pusat bangun.
Program menerima masukan nilai integer positif N dari pengguna yang merepresentasikan ukuran skala jam pasir (skala angka tertinggi pada puncak dan alas).
Struktur Pembentukan Pola (3 Fase + 2 Bingkai Horizontal):

**Kode Jawaban :**

```python
print ("PROGRAM JAM PASIR KRISTAL PALINDROMIK")
n_3015 = int (input ("Masukkan ukuran skala jam pasir (N): "))

# Bingkai atas
print ("#", end ="")
for b_3015 in range (4* n_3015 +5):
    print ("=", end ="")
print ("#", end = "")
print ()

# fase 1: jam pasir atas
for baris_3015 in range (n_3015, 0, -1):
    print ("|", end="")
    print (" ", end="")

    # spasi penyeimbang kiri
    for sp_kiri_3015 in range (2 * (n_3015 - baris_3015)):
        print (" ", end="")

     # deret angka mundur
    for angka_3015 in range (baris_3015, 0, -1):
        print (angka_3015, end=" ")

    # poros kristal
    print ("<*>", end="")

    # deret angka maju
    for angka_3015 in range (1, baris_3015 + 1):
        print (" ", end="")
        print (angka_3015, end="")

    # spasi penyeimbang kanan
    for sp_kanan_3015 in range (2 * (n_3015 - baris_3015)):
        print (" ", end="")

    print (" ", end="")
    print ("|", end="")
    print ()

# Fase 2: Poros titik pusat
print ("|", end="")
for sp_kiri_3015 in range (2 * n_3015 + 1):
    print (" ", end="")
print ("<*>", end="")
for sp_kanan_3015 in range (2 * n_3015 + 1):
    print (" ", end="")
print ("|", end="")
print ()

# Fase 3: Jam pasir bawah
for baris_3015 in range (1, n_3015 + 1):
    print ("|", end="")
    print (" ", end="")

    # spasi penyeimbang kiri
    for sp_kiri_3015 in range (2 * (n_3015 - baris_3015)):
        print (" ", end="")

    # deret angka mundur
    for angka_3015 in range (baris_3015, 0, -1):
        print (angka_3015, end=" ")

    # poros kristal
    print ("<*>", end="")

    # deret angka maju
    for angka_3015 in range (1, baris_3015 + 1):
        print (" ", end="")
        print (angka_3015, end="")

    # spasi penyeimbang kanan
    for sp_kanan_3015 in range (2 * (n_3015 - baris_3015)):
        print (" ", end="")

    print (" ", end="")
    print ("|", end="")
    print ()

# Bingkai bawah
print ("#", end="")
for b_3015 in range (4 * n_3015 + 5):
    print ("=", end="")
print ("#", end="")
print ()
```
