Program memakai if-elif-else dengan rentang: >= 81 (A), 70-80 (B), 60-69 (C),

```python
umur_3015 = int (input ("Input umur anda: "))
sim_3015 = input ("Apakah Anda sudah Punya Sim C (y/t): ")[0]

if umur_3015 >= 17 and sim_3015 == 'y':
    print ("Anda Sudah dewasa dan boleh bawa motor")

elif umur_3015 >= 17 and sim_3015 != 'y':
    print ("Anda Sudah dewasa tetapi tidak boleh bawa motor")

elif umur_3015 < 17 and sim_3015 == 'y':
    print ("Anda Belum cukup umur punya SIM")

else:
    print ("Anda Belum cukup umur dan tidak boleh bawa motor")
print ("Program Selesai")
```

```python
bulan_3015 = int(input("Masukkan angka bulan (1 - 12): "))

match bulan_3015:
    case 1:
        print ("Januari")

    case 2:
        print ("Februari")

    case 3:
        print ("Maret")

    case 4:
        print ("April")

    case 5:
        print ("Mei")

    case 6:
        print ("Juni")

    case 7:
        print ("Juli")

    case 8:
        print ("Agustus")

    case 9:
        print ("September")

    case 10:
        print ("Oktober")

    case 11:
        print ("November")

    case 12:
        print ("Desember")

    case _:
        print ("Angks tidak valid")

```

PENJELASAN PROGRAM: KONVERSI ANGKA BULAN MENJADI NAMA BULAN (match-case)

1. TUJUAN PROGRAM
Program ini meminta pengguna memasukkan angka bulan (1 sampai 12), lalu
menampilkan nama bulan yang sesuai. Jika angka di luar 1-12, program
menampilkan pesan "Angks tidak valid".

2. INPUT DATA

bulan_3015 = int(input("Masukkan angka bulan (1 - 12): "))
- input() menampilkan teks lalu menunggu pengguna mengetik.
- Hasil input() selalu bertipe string, jadi dibungkus int() agar menjadi
  bilangan bulat dan cocok dengan pola case yang berupa angka.
- Hasilnya disimpan di variabel bulan_3015.

3. STRUKTUR match-case

match bulan_3015:
- Statement match membandingkan nilai bulan_3015 dengan pola pada setiap case
  dari atas ke bawah.
- Begitu ada satu case yang cocok, blok case itu dijalankan, lalu seluruh
  match selesai. Case sisanya dilewati.
- Tidak perlu break seperti switch-case di bahasa C atau Java.
- match-case tersedia mulai Python 3.10. Pada versi lebih lama akan terjadi
  SyntaxError.

case 1: sampai case 12:
- Setiap case berisi satu nilai angka yang dicocokkan dengan bulan_3015.
- Contoh: jika bulan_3015 bernilai 3, maka yang dijalankan adalah
  case 3 -> print("Maret").

case _:
- Tanda _ adalah wildcard (pola penampung) yang cocok dengan nilai apa pun.
- Berfungsi seperti else pada if-elif-else, yaitu berjalan jika tidak ada
  case sebelumnya yang cocok.
- Harus diletakkan paling akhir.

4. DAFTAR NILAI DAN OUTPUT

5     | Mei

9     | September

lainnya | Angks tidak valid

5. CONTOH EKSEKUSI

Masukkan angka bulan (1 - 12): 0   -> "Angks tidak valid"
Masukkan angka bulan (1 - 12): 15  -> "Angks tidak valid"
Masukkan angka bulan (1 - 12): -3  -> "Angks tidak valid"

6. PERBANDINGAN DENGAN if-elif-else
- match-case lebih rapi dan mudah dibaca untuk banyak pilihan nilai tetap.
- Dengan if-elif-else, setiap kondisi harus ditulis lengkap, misalnya
  elif bulan_3015 == 2:
- Hasil program sama, hanya penulisannya yang berbeda.

7. CATATAN / KELEMAHAN PROGRAM
- Ada salah ketik pada pesan error: "Angks tidak valid". Seharusnya
  "Angka tidak valid".
- Jika pengguna mengisi huruf atau desimal (misalnya "abc" atau "2.5"),
  int() menyebabkan ValueError sebelum match dijalankan.
- Program membutuhkan Python 3.10 atau lebih baru.
- Ada 12 case terpisah yang berulang. Alternatif yang lebih singkat adalah
  memakai list nama bulan dan mengaksesnya dengan indeks.

```python
umur_3015 = int (input ("Input umur anda: "))
sim_3015 = input ("Apakah Anda sudah Punya Sim C (y/t):")[0]

if umur_3015 >= 17 and sim_3015 == 'y':
    print ("Anda Sudah dewasa dan boleh bawa motor")

if umur_3015 >= 17 and sim_3015 != 'y':
    print ("Anda Sudah dewasa tetapi tidak boleh bawa motor")

if umur_3015 < 17 and sim_3015 == 'y':
    print ("Anda Belum cukup umur punya SIM")

if umur_3015 < 17 and sim_3015 != 'y':
    print ("Anda Belum cukup umur bawa motor")
```

2. INPUT DATA

- Membandingkan nilai paket_3015 dengan tiap case dari atas ke bawah.
- Begitu ada case yang cocok, blok itu dijalankan lalu match selesai.
  Tidak perlu break.
- match-case tersedia mulai Python 3.10.

case 1 sampai case 5:
- Tiap case mengisi dua variabel:
  nama_wahana_3015  : nama paket
  harga_satuan_3015 : harga per tiket
- Daftar paket:
  1 -> Wahana Safari Rimba          Rp 50.000
  2 -> Wahana Arung Jeram           Rp 75.000
  3 -> Wahana Motor ATV Ekstrim     Rp 120.000
  4 -> Wahana Roller Coaster Kilat  Rp 100.000
  5 -> Wahana All-Access VIP        Rp 220.000

case _:
- Wildcard: cocok dengan nilai apa pun yang tidak ada di case 1-5.
- Mencetak "Paket wahana tidak valid!" lalu sys.exit() menghentikan program.
- Karena program berhenti di sini, variabel harga_satuan_3015 tidak pernah
  dipakai dalam keadaan belum terisi.

7. INPUT JUMLAH TIKET DAN VALIDASI (if tunggal)

jumlah_tiket_3015 = int(input("Masukkan jumlah tiket           : "))

if jumlah_tiket_3015 <= 0:
    print("Peringatan: kuota tiket tidak valid!")
    sys.exit()
- if tunggal tanpa else. Jika jumlah tiket 0 atau negatif, program mencetak
  peringatan lalu berhenti.
- Jika jumlah tiket valid (>= 1), blok ini dilewati dan program lanjut.

8. INPUT MEMBER DAN PROMO

is_member_3015 = input("Apakah Anda member? (y/t)       : ").strip().lower()
kode_promo_valid_3015 = input("Apakah kode promo valid? (y/t)  : ").strip().lower()
- Sama seperti input SIM, tetapi tanpa [:1], sehingga yang tersimpan adalah
  seluruh jawaban (contoh: "y", "ya", "t", "tidak").
- Nanti diperiksa dengan operator in terhadap list ['y', 'ya'].
- Jawaban di luar 'y' dan 'ya' (misalnya "iya" atau "yes") dianggap tidak.

9. VALIDASI IZIN KENDALI (satu rantai if-elif-else)

print("\n--- KELAYAKAN PENGENDARA WAHANA ---")
- Judul bagian kelayakan.

Rantai kondisi diperiksa dari atas ke bawah dan berhenti di kondisi
pertama yang benar:

Kondisi 1:
if paket_3015 == 3 and umur_3015 >= 17 and sim_3015 == 'y':
- Paket ATV, umur 17 ke atas, punya SIM.
- Hasil: "Anda sudah dewasa dan boleh mengendarai ATV sendiri."

Kondisi 2:
elif paket_3015 == 3 and umur_3015 >= 17 and sim_3015 != 'y':
- Paket ATV, umur 17 ke atas, tidak punya SIM.
- Hasil: "... wajib didampingi instruktur)."

Kondisi 3:
elif paket_3015 == 3 and umur_3015 < 17 and sim_3015 == 'y':
- Paket ATV, umur di bawah 17, mengaku punya SIM.
- Hasil: "Identitas tidak valid: Belum cukup umur memiliki SIM."

Kondisi 4:
elif paket_3015 == 3:
- Paket ATV, dan karena tiga kondisi sebelumnya salah, berarti umur di bawah
  17 dan tidak punya SIM. Tidak perlu menulis ulang syaratnya.
- Hasil: "Anda belum cukup umur dan tidak boleh bawa motor ATV."

Kondisi 5:
elif umur_3015 >= 10:
- Hanya bisa tercapai jika paket BUKAN 3 (karena semua kasus paket 3 sudah
  tertangkap kondisi 1-4).
- Hasil: "Umur memenuhi syarat untuk wahana ini."

Kondisi 6:
else:
- Paket bukan 3 dan umur di bawah 10.
- Hasil: "Maaf, umur minimal untuk wahana ini adalah 10 tahun."

Kenapa paket_3015 == 3 diulang pada tiap kondisi?
- Aturan soal melarang nested-if (if di dalam if). Jika ditulis
  "if paket == 3:" lalu if-elif-else di dalamnya, itu nested-if.
- Solusinya, satu rantai datar dengan syarat paket == 3 digabung memakai and.

print("Status Akses:", status_akses_3015)
- Mencetak hasil pemeriksaan.
- Tanda koma pada print membuat Python memberi satu spasi antar bagian.

10. DISKON BERTINGKAT (multi-if terpisah)

subtotal_3015 = harga_satuan_3015 * jumlah_tiket_3015
- Subtotal = harga satuan dikali jumlah tiket.

total_diskon_persen_3015 = 0
- Nilai awal diskon 0%. Nilainya bertambah sesuai kondisi yang terpenuhi.

Empat if terpisah (bukan elif):
- if subtotal_3015 >= 200000        -> tambah 10 (diskon belanja besar)
- if is_member_3015 in ['y', 'ya']  -> tambah 5  (diskon member)
- if kode_promo_valid_3015 in ['y', 'ya'] -> tambah 15 (diskon voucher)
- if jumlah_tiket_3015 >= 5         -> tambah 5  (diskon rombongan)

Catatan penting:
- Memakai if terpisah membuat SEMUA kondisi selalu diperiksa, sehingga
  diskon bisa ditumpuk. Jika memakai elif, hanya satu diskon yang berlaku.
- Operator in memeriksa apakah nilai ada di dalam list, hasilnya True/False.
- += adalah tambah lalu simpan: x += 10 sama dengan x = x + 10.
- Diskon maksimal adalah 10 + 5 + 15 + 5 = 35%.
- Batas belanja besar memakai >= sehingga subtotal tepat 200.000 sudah
  mendapat diskon.

11. TOTAL BAYAR DAN CATATAN LAYANAN (if-else)

nominal_diskon_3015 = subtotal_3015 * (total_diskon_persen_3015 / 100)
- Persen dibagi 100 menjadi pecahan, dikali subtotal.
  Contoh: 30% dari 240.000 = 240000 * 0.30 = 72000.

total_bayar_3015 = subtotal_3015 - nominal_diskon_3015
- Total bayar = subtotal dikurangi nominal diskon.

if total_bayar_3015 > 300000:
    catatan_3015 = "Selamat! Anda berhak mendapatkan Souvenir Gratis."
else:
    catatan_3015 = "Terima kasih telah berkunjung."
- Jika total bayar LEBIH BESAR dari 300.000 (bukan sama dengan), pengunjung
  mendapat souvenir. Selain itu, ucapan terima kasih.
- Pesan disimpan dulu ke variabel, lalu dicetak di bagian output.

12. OUTPUT

print("\n--- Rincian Pembayaran ---")
print(f"Subtotal Belanja : Rp {subtotal_3015:,.0f}")
print(f"Total Diskon     : {total_diskon_persen_3015}% (Rp {nominal_diskon_3015:,.0f})")
print(f"Total Bayar      : Rp {total_bayar_3015:,.0f}")
print(f"Catatan Layanan  : {catatan_3015}")
print("Program Selesai")
- Awalan f membuat f-string, sehingga nilai variabel di dalam { } langsung
  dimasukkan ke teks.
- Format :,.0f berarti:
  ,   = pemisah ribuan
  .0f = float tanpa angka di belakang koma (dibulatkan saat ditampilkan)
- print("Program Selesai") berada di luar semua blok, jadi selalu dijalankan
  jika program tidak dihentikan oleh sys.exit().
