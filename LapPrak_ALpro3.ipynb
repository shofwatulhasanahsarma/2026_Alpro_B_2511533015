## 2.1 Dasar Teori
1. Operator Aritmatika

Operator aritmatika digunakan untuk melakukan operasi matematis dasar seperti penjumlahan (+), pengurangan (-), perkalian (*), pembagian (/), pembagian bulat/floor division (//), sisa bagi/modulus (%), dan perpangkatan (**).

2. Operator Assignment

Operator assignment digunakan untuk memberikan nilai ke suatu variabel. Selain assignment biasa (=), Python menyediakan compound assignment seperti += yang setara dengan x = x + y, sehingga kode menjadi lebih ringkas.

3. Operator Bitwise

Operator bitwise bekerja pada level bit dari representasi biner suatu bilangan bulat. AND (&) menghasilkan 1 jika kedua bit bernilai 1, OR (|) menghasilkan 1 jika salah satu bit bernilai 1, XOR (^) menghasilkan 1 jika kedua bit berbeda, NOT (~) membalik seluruh bit, sedangkan shift kiri (<<) dan shift kanan (>>) menggeser posisi bit.

4. Operator Keanggotaan

Operator in dan not in digunakan untuk memeriksa apakah suatu nilai terdapat (atau tidak terdapat) di dalam struktur data seperti list, tuple, atau string.

5. Operator Identitas

Operator is dan is not digunakan untuk membandingkan identitas (alamat memori) dua objek, berbeda dengan == yang membandingkan nilai/isi objek. Dua objek dapat memiliki nilai yang sama (== bernilai True) namun bukan objek yang sama secara identitas (is bernilai False).

6. Operator Logika
Operator logika (and, or, not) digunakan untuk mengombinasikan atau membalik nilai boolean, umumnya digunakan dalam struktur kontrol percabangan. Operator ^ pada tipe boolean berfungsi sebagai XOR logika.

7. Operator Perbandingan

Operator perbandingan digunakan untuk membandingkan dua nilai dan menghasilkan output boolean (True/False). Python juga mendukung chained comparison seperti 0 < x < 100 yang setara dengan (0 < x) and (x < 100).

```python
angka1_3015 = int(input("Input angka-1: "))
angka2_3015 = int(input("Input angka-2: "))

#Penjumlahan
hasil_3015 = angka1_3015 + angka2_3015
print("\nOperator Penjumlahan")
print("Hasil =", hasil_3015)

#Pengurangan
hasil_3015 = angka1_3015 - angka2_3015
print("\nOperator Pengurangan")
print("Hasil =", hasil_3015)

#Perkalian
hasil_3015 = angka1_3015 * angka2_3015
print("\nOperator Perkalian")
print("Hasil =", hasil_3015)

#Pembagian, pembagian bulat, sisa bagi
if angka2_3015 !=0:
    hasil_3015 = angka1_3015 / angka2_3015
    print("\nOperator Pembagian")
    print("Hasil =", hasil_3015)

    hasil_3015 = angka1_3015 // angka2_3015
    print("\nOperator Pembagian Bulat")
    print("Hasil =", hasil_3015)

    hasil_3015 = angka1_3015 % angka2_3015
    print("\nOperator Sisa Bagi")
    print("Hasil =", hasil_3015)
else:
    print("Angka kedua tidak boleh bernilai 0.")

#Pangkat
hasil_3015 = angka1_3015 ** angka2_3015
print("\nOperator Pangkat")
print("Hasil =", hasil_3015)
```

Program ini merupakan implementasi operator aritmatika dalam Python yang meminta dua buah angka dari pengguna, kemudian melakukan berbagai operasi matematis terhadap kedua angka tersebut.

1. Input Data
Baris pertama dan kedua meminta pengguna memasukkan dua buah angka (angka1_3015 dan angka2_3015) melalui fungsi input(), yang kemudian dikonversi menjadi tipe data integer menggunakan int().

2. Operator Penjumlahan (+)
Menjumlahkan angka1_3015 dan angka2_3015, hasilnya disimpan dalam variabel hasil_3015 lalu ditampilkan ke layar.

3. Operator Pengurangan (-)
Mengurangkan angka1_3015 dengan angka2_3015 menggunakan operator -, kemudian hasilnya ditampilkan.

4. Operator Perkalian (*)
Mengalikan kedua angka menggunakan operator *, lalu hasil perkalian ditampilkan ke layar.

5. Operator Pembagian, Pembagian Bulat, dan Sisa Bagi
Bagian ini diawali dengan kondisi if angka2_3015 != 0 untuk mencegah terjadinya error ZeroDivisionError yang muncul apabila pembagi bernilai nol. Jika angka2_3015 tidak sama dengan nol, maka dijalankan tiga jenis operasi pembagian:

Pembagian biasa (/) — menghasilkan nilai desimal (float).
Pembagian bulat (//) — menghasilkan hasil bagi yang dibulatkan ke bawah (integer).
Sisa bagi/modulus (%) — menghasilkan sisa dari hasil pembagian.

Jika angka2_3015 bernilai nol, program akan menampilkan pesan "Angka kedua tidak boleh bernilai 0." melalui blok else, sehingga ketiga operasi pembagian di atas tidak dijalankan.

6. Operator Pangkat ()**
Menghitung angka1_3015 dipangkatkan dengan angka2_3015 menggunakan operator **, lalu hasilnya ditampilkan. Operasi ini diletakkan di luar blok if-else sehingga tetap dijalankan meskipun angka2_3015 bernilai nol (karena pemangkatan dengan nol tidak menyebabkan error).

```python
angka1_3015 = int(input("Input angka-1: "))
angka2_3015 = int(input("Input angka-2: "))

print("\nNilai awal angka1_3015 =", angka1_3015)
print("Nilai angka2_3015 =", angka2_3015)

#Assignment biasa
hasil_3015 = angka1_3015
print("\nOperator Assignment Biasa (=)")
print("Hasil =", hasil_3015)

#Assignment penjumlahan
hasil_3015 = angka1_3015
hasil_3015 += angka2_3015
print("\nOperator Assignment Penjumlahan (+=)")
print("Hasil =", hasil_3015)

#Assignment pengurangan
hasil_3015 = angka1_3015
hasil_3015 -= angka2_3015
print("\nOperator Assignment Pengurangan (-=)")
print("Hasil =", hasil_3015)

#Assignment perkalian
hasil_3015 = angka1_3015
hasil_3015 *= angka2_3015
print("\nOperator Assignment Perkalian (*=)")
print("Hasil =", hasil_3015)

#Assignment pembagian, pembagian bulat, sisa bagi
if angka2_3015 !=0:
    hasil_3015 = angka1_3015
    hasil_3015 /= angka2_3015
    print("\nOperator Assignment Pembagian (/=)")
    print("Hasil =", hasil_3015)
    #Operator tambahan
    hasil_3015 = angka1_3015
    hasil_3015 //= angka2_3015
    print("\nOperator Assignment Pembagian Bulat (//=)")
    print("Hasil =", hasil_3015)

    hasil_3015 = angka1_3015
    hasil_3015 %= angka2_3015
    print("\nOperator Assignment Sisa Bagi (%=)")
    print("Hasil =", hasil_3015)
else:
    print("\nPembagian tidak dapat dilakukan.")
    print("Angka kedua tidak boleh bernilai 0.")

#Operator tambahan: Assignment perpangkatan
hasil_3015 = angka1_3015
hasil_3015 **= angka2_3015
print("\nOperator Assignment Perpangkatan (**=)")
print("Hasil =", hasil_3015)
```

Program ini mendemonstrasikan operator assignment (penugasan) dalam Python, yaitu operator yang digunakan untuk memberikan atau memperbarui nilai suatu variabel, termasuk bentuk gabungan (compound assignment) yang menyingkat operasi aritmatika sekaligus penugasan dalam satu langkah.

1. Input Data
Program meminta dua angka dari pengguna (angka1_3015 dan angka2_3015), lalu menampilkan nilai awal keduanya sebagai referensi sebelum dilakukan operasi assignment.

2. Assignment Biasa (=)
hasil_3015 = angka1_3015 merupakan operator assignment paling dasar, yaitu memberikan nilai angka1_3015 langsung ke variabel hasil_3015 tanpa operasi tambahan.

3. Assignment Penjumlahan (+=)
hasil_3015 += angka2_3015 setara dengan hasil_3015 = hasil_3015 + angka2_3015. Karena hasil_3015 sebelumnya di-reset ke angka1_3015, maka hasil akhirnya sama dengan angka1_3015 + angka2_3015.

4. Assignment Pengurangan (-=)
Sama seperti di atas, namun menggunakan operasi pengurangan: hasil_3015 -= angka2_3015 setara dengan hasil_3015 = hasil_3015 - angka2_3015.

5. Assignment Perkalian (*=)
hasil_3015 *= angka2_3015 setara dengan hasil_3015 = hasil_3015 * angka2_3015, menghasilkan perkalian antara angka1_3015 dan angka2_3015.

6. Assignment Pembagian, Pembagian Bulat, dan Sisa Bagi
Bagian ini dibungkus dalam kondisi if angka2_3015 != 0 untuk mencegah ZeroDivisionError. Jika syarat terpenuhi, dijalankan tiga operator:

/= — pembagian biasa, hasil berupa desimal (float).
//= — pembagian bulat, hasil dibulatkan ke bawah.
%= — sisa bagi (modulus).

Setiap operator selalu didahului dengan me-reset hasil_3015 = angka1_3015 terlebih dahulu, sehingga setiap operasi dihitung dari nilai awal angka1_3015, bukan hasil akumulasi dari operasi sebelumnya. Jika angka2_3015 bernilai nol, blok else akan menampilkan pesan bahwa pembagian tidak dapat dilakukan.

7. Assignment Perpangkatan (=)**
hasil_3015 **= angka2_3015 setara dengan hasil_3015 = hasil_3015 ** angka2_3015, yaitu memangkatkan angka1_3015 dengan angka2_3015. Operasi ini diletakkan di luar blok if-else karena pemangkatan tetap valid meskipun angka2_3015 bernilai nol.

```python
print("\n=============================================")
print("3. OPERATOR BITWISE")
print("=============================================")

angka1_3015 = int(input("Masukkkan angka bitwise-1: "))
angka2_3015 = int(input("Masukkkan angka bitwise-2: "))

print("\nAngka dalam bentuk desimal dan biner")
print("Angka1_3015 =", angka1_3015, " | biner =", bin(angka1_3015))
print("Angka2_3015 =", angka2_3015, " | biner =", bin(angka2_3015))

#Bitwise AND
hasil_3015 = angka1_3015 & angka2_3015
print("\nOperator Bitwise AND (&)")
print(angka1_3015, "&", angka2_3015, "=", hasil_3015)
print("Biner hasil =", bin(hasil_3015))
print("Biner hasil (8 bit) =", format(hasil_3015, "08b"))

#Bitwise OR
hasil_3015 = angka1_3015 | angka2_3015
print("\nOperator Bitwise OR (|)")
print(angka1_3015, "|", angka2_3015, "=", hasil_3015)
print("Biner hasil =", bin(hasil_3015))
print("Biner hasil (8 bit) =", format(hasil_3015, "08b"))

#Bitwise XOR
hasil_3015 = angka1_3015 ^ angka2_3015
print("\nOperator Bitwise XOR (^)")
print(angka1_3015, "^", angka2_3015, "=", hasil_3015)
print("Biner hasil =", bin(hasil_3015))
print("Biner hasil (8 bit) =", format(hasil_3015, "08b"))

#Bitwise NOT
hasil_3015 = ~angka1_3015
print("\nOperator Bitwise NOT (~)")
print("~", angka1_3015, "=", hasil_3015)
print("Biner hasil =", bin(hasil_3015))
print("Biner hasil (8 bit) =", format(hasil_3015, "08b"))

#Bitwise geser kiri
jumlah_geser_3015 = int(input("\nMasukkan jumlah pergeseran bit: "))
hasil_3015 = angka1_3015 << jumlah_geser_3015
print("\nOperator Bitwise geser kiri (<<)")
print(angka1_3015, "<<", jumlah_geser_3015, "=", hasil_3015)
print("Biner hasil =", bin(hasil_3015))
print("Biner hasil (8 bit) =", format(hasil_3015, "08b"))

#Bitwise geser kanan
hasil_3015 = angka1_3015 >> jumlah_geser_3015
print("\nOperator Bitwise geser kanan (>>)")
print(angka1_3015, ">>", jumlah_geser_3015, "=", hasil_3015)
print("Biner hasil =", bin(hasil_3015))
print("Biner hasil (8 bit) =", format(hasil_3015, "08b"))
```

Program ini mendemonstrasikan operator bitwise dalam Python, yaitu operator yang bekerja langsung pada representasi biner (bit) dari suatu bilangan bulat.

1. Input dan Tampilan Biner
Program meminta dua angka dari pengguna, lalu menampilkan representasi biner masing-masing menggunakan fungsi bin(), yang menghasilkan string biner dengan awalan 0b.

2. Bitwise AND (&)
Operator & membandingkan setiap bit dari kedua angka; hasilnya bernilai 1 hanya jika kedua bit yang dibandingkan sama-sama 1, selain itu 0. Hasilnya ditampilkan dalam bentuk desimal, biner standar (bin()), dan biner 8-bit menggunakan format(hasil_3015, "08b") yang menambahkan angka nol di depan agar selalu tampil 8 digit.

3. Bitwise OR (|)
Operator | menghasilkan bit 1 jika salah satu atau kedua bit yang dibandingkan bernilai 1. Hasil hanya 0 jika kedua bit sama-sama 0.

4. Bitwise XOR (^)
Operator ^ menghasilkan bit 1 jika kedua bit yang dibandingkan berbeda nilai, dan 0 jika bit tersebut sama.

5. Bitwise NOT (~)
Operator ~ membalik seluruh bit dari angka1_3015 (komplemen satu). Karena Python merepresentasikan integer negatif menggunakan two's complement, hasil dari ~x secara matematis setara dengan -(x + 1).

6. Bitwise Geser Kiri (<<)
Program meminta input tambahan berupa jumlah pergeseran bit (jumlah_geser_3015). Operator << menggeser seluruh bit angka1_3015 ke kiri sejumlah posisi yang ditentukan, dengan bit kosong di sebelah kanan diisi 0. Secara matematis, ini setara dengan mengalikan angka dengan 2 sebanyak jumlah pergeseran (x * 2^n).

7. Bitwise Geser Kanan (>>)
Operator >> menggeser seluruh bit angka1_3015 ke kanan sejumlah posisi yang sama (jumlah_geser_3015), yang setara dengan pembagian bulat oleh 2 sebanyak jumlah pergeseran (x // 2^n).

Catatan tambahan: fungsi format(hasil_3015, "08b") akan menampilkan hasil kurang akurat untuk bilangan negatif (seperti hasil dari ~), karena Python tidak membatasi representasi bit pada lebar tetap (8-bit) secara native — tampilan tersebut lebih cocok digunakan untuk bilangan non-negatif kecil.

```python
print("==========================================")
print("1. OPERATOR KEANGGOTAAN")
print("==========================================")

#Input beberapa data yang dipisahkan dengan koma
input_data_3015 = input("Masukkan beberapa angka, pisahkan dengan koma:")

#Mengubah input menjadi list Integer
data_3015 = [int(angka.strip()) for angka in input_data_3015.split(",")]

nilai_dicari_3015 = int(input("Masukkan nilai yang dicari: "))

#Operator in
hasil_3015 = nilai_dicari_3015 in data_3015
print("\nOperator Keanggotaan in")
print(nilai_dicari_3015, "in", data_3015, "=", hasil_3015)

#Operator not in
hasil_3015 = nilai_dicari_3015 not in data_3015
print("\nOperator Keanggotaan not in")
print(nilai_dicari_3015, "not in", data_3015, "=", hasil_3015)

print("\n==========================================")
print("2. OPERATOR IDENTITAS")
print("==========================================")

#Objkek1 menggunakan list dari input pengguna
objek1_3015 = data_3015

#Objek2 merujuk pada objek yang sama dengan objek1_3015
objek2_3015 = objek1_3015

#Objek3 memiliki isi sama, tetapi merupakan objek baru
objek3_3015 = data_3015.copy()

print("Objek1 =", objek1_3015)
print("Objek2 =", objek2_3015)
print("Objek3 =", objek3_3015)

#Operator is
hasil_3015 = objek1_3015 is objek2_3015
print("\nOperator Identitas is")
print("objek1_3015 is objek2_3015 =", hasil_3015)

#Operator is not
hasil_3015 = objek1_3015 is not objek3_3015
print("\nOperator Identitas is not")
print("objek1_3015 is not objek3_3015 =", hasil_3015)

#Membandingkan identitas dan nilai
print("\nPerbandingan identitas dan nilai")
print("objek1_3015 is objek3_3015 =", objek1_3015 is objek3_3015)
print("objek1_3015 == objek3_3015 =", objek1_3015 == objek3_3015)
```

Program ini mendemonstrasikan dua kelompok operator dalam Python: operator keanggotaan (in, not in) dan operator identitas (is, is not).

1. Operator Keanggotaan

Input dan Pengolahan Data
Program meminta pengguna memasukkan beberapa angka yang dipisahkan koma (misalnya 1, 2, 3, 4). String input tersebut kemudian diproses menggunakan list comprehension:

python
data_3015 = [int(angka.strip()) for angka in input_data_3015.split(",")]

Baris ini memecah string berdasarkan tanda koma (split(",")), menghapus spasi di sekitar setiap elemen (strip()), lalu mengonversi setiap elemen menjadi integer (int()), sehingga menghasilkan sebuah list bilangan bulat.

Operator in
nilai_dicari_3015 in data_3015 akan bernilai True jika nilai yang dicari ditemukan di dalam list data_3015, dan False jika tidak ditemukan.

Operator not in
Kebalikan dari in — nilai_dicari_3015 not in data_3015 akan bernilai True jika nilai tersebut tidak ditemukan dalam list.

2. Operator Identitas

Pembuatan Tiga Objek

objek1_3015 = data_3015 → objek1_3015 tidak membuat list baru, melainkan hanya menyalin referensi (alamat memori) dari data_3015. Jadi keduanya menunjuk ke objek yang sama persis di memori.
objek2_3015 = objek1_3015 → Sama seperti di atas, objek2_3015 juga hanya menyalin referensi dari objek1_3015, sehingga ketiganya (data_3015, objek1_3015, objek2_3015) menunjuk ke objek list yang sama.
objek3_3015 = data_3015.copy() → Method .copy() membuat salinan baru (objek independen) dari data_3015 dengan isi yang identik, namun disimpan pada alamat memori yang berbeda.

Operator is
objek1_3015 is objek2_3015 mengecek apakah kedua variabel menunjuk ke objek yang benar-benar sama di memori (bukan sekadar nilai yang sama). Karena objek2_3015 adalah referensi langsung dari objek1_3015, hasilnya pasti True.

Operator is not
objek1_3015 is not objek3_3015 mengecek kebalikannya. Karena objek3_3015 adalah hasil .copy() (objek baru), meskipun isinya sama, keduanya bukan objek yang sama, sehingga hasilnya True.

Perbandingan Identitas vs Nilai
Dua baris terakhir menegaskan perbedaan mendasar antara is dan ==:

objek1_3015 is objek3_3015 → False, karena keduanya adalah objek yang berbeda di memori (meski isinya sama).
objek1_3015 == objek3_3015 → True, karena == hanya membandingkan isi/nilai dari kedua list, bukan lokasi memorinya.

Program menerima data transaksi dan informasi pelanggan, kemudian menggunakan berbagai jenis operator Python untuk menentukan total pembayaran, status diskon, kelayakan promo, serta hak akses pelanggan. ```python

```python
"""
=====================================================================
 STUDI KASUS: SISTEM SIMULASI TRANSAKSI DAN VALIDASI AKSES TOKO
=====================================================================
Catatan penting:
- Semua nama variabel diakhiri dengan 4 digit terakhir NIM.
- Pada contoh ini digunakan akhiran "1234". GANTI seluruh akhiran
  "_3015" pada kode ini dengan 4 digit terakhir NIM Anda sendiri
  sebelum dikumpulkan (misalnya cari-ganti "_3015" -> "_5678").
=====================================================================
"""

# =====================================================================
# 1. INPUT DATA PELANGGAN DAN TRANSAKSI
# =====================================================================
print("=== SISTEM TRANSAKSI TOKO ===")
print()

nama_3015 = input("Masukkan Nama Pelanggan : ")                      # tipe data: str
status_3015 = input("Masukkan Status Pelanggan (member/nonmember) : ").strip().lower()  # str
total_belanja_3015 = float(input("Masukkan Total Belanja : "))        # tipe data: float
jumlah_barang_3015 = int(input("Masukkan Jumlah Barang : "))          # tipe data: int
kode_promo_3015 = input("Masukkan Kode Promo : ").strip().upper()     # str

# Daftar kode promo yang tersedia di toko
daftar_promo_3015 = ["HEMAT10", "HEMAT20", "GRATISONGKIR"]

# =====================================================================
# 2. OPERATOR PERBANDINGAN (comparison operator) -> menghasilkan True/False
# =====================================================================
syarat_belanja_3015 = total_belanja_3015 >= 200000   # apakah belanja >= Rp200.000
syarat_barang_3015 = jumlah_barang_3015 >= 3         # apakah jumlah barang >= 3
is_member_3015 = status_3015 == "member"             # apakah pelanggan member

# =====================================================================
# 3. OPERATOR LOGIKA (and, or, not) -> gabungan minimal dua kondisi
# =====================================================================
# Diskon member khusus: harus member DAN memenuhi syarat belanja minimum
diskon_member_3015 = is_member_3015 and syarat_belanja_3015

# Promo tambahan: memenuhi syarat jumlah barang ATAU sudah dapat diskon member
promo_tambahan_3015 = syarat_barang_3015 or diskon_member_3015

# Menandai pelanggan yang BUKAN member menggunakan operator not
bukan_member_3015 = not is_member_3015

# =====================================================================
# 4. OPERATOR KEANGGOTAAN (membership: in / not in)
# =====================================================================
promo_tersedia_3015 = kode_promo_3015 in daftar_promo_3015        # True jika kode ada di daftar
promo_tidak_tersedia_3015 = kode_promo_3015 not in daftar_promo_3015  # kebalikannya

# Pelanggan mendapatkan promo jika kode promo tersedia DAN memenuhi salah satu syarat
mendapatkan_promo_3015 = promo_tersedia_3015 and (diskon_member_3015 or syarat_barang_3015)

# =====================================================================
# 5. OPERATOR ARITMATIKA (+, -, *, /, %)
# =====================================================================
persen_diskon_3015 = 0.1 if diskon_member_3015 else 0          # persentase diskon (10%)
diskon_rupiah_3015 = total_belanja_3015 * persen_diskon_3015    # perkalian -> nilai diskon
total_pembayaran_3015 = total_belanja_3015 - diskon_rupiah_3015 # pengurangan -> total setelah diskon
rata_rata_harga_3015 = total_belanja_3015 / jumlah_barang_3015  # pembagian -> harga rata-rata barang
sisa_bagi_3015 = jumlah_barang_3015 % 3                          # modulus -> sisa pembagian jumlah barang oleh 3

# =====================================================================
# 6. OPERATOR PENUGASAN (assignment & augmented assignment)
# =====================================================================
poin_3015 = 0
poin_3015 += jumlah_barang_3015 * 10     # augmented assignment (+=): poin bertambah dari jumlah barang

if mendapatkan_promo_3015:
    total_pembayaran_3015 -= 5000        # augmented assignment (-=): potongan tambahan jika dapat promo

if is_member_3015:
    poin_3015 *= 2                       # augmented assignment (*=): poin member digandakan

# =====================================================================
# 7. OPERATOR IDENTITAS (is / is not)
# =====================================================================
# Dua objek list dengan ISI SAMA tetapi merupakan objek berbeda di memori
list_a_3015 = ["HEMAT10"]
list_b_3015 = ["HEMAT10"]

cek_is_3015 = list_a_3015 is list_b_3015           # False -> objek berbeda walau isinya sama
cek_is_not_3015 = list_a_3015 is not list_b_3015   # True  -> karena identitas objek berbeda
cek_nilai_sama_3015 = list_a_3015 == list_b_3015   # True  -> == membandingkan ISI, bukan identitas

# =====================================================================
# 8. OPERATOR BITWISE (&, |, ^, <<)
# =====================================================================
# Representasi bit sesuai tabel:
# 0001 = member | 0010 = belanja>=200000 | 0100 = barang>=3 | 1000 = promo tersedia
bit_member_3015 = 0b0001 if is_member_3015 else 0
bit_belanja_3015 = 0b0010 if syarat_belanja_3015 else 0
bit_barang_3015 = 0b0100 if syarat_barang_3015 else 0
bit_promo_3015 = 0b1000 if promo_tersedia_3015 else 0

# OR (|) -> menggabungkan seluruh kondisi menjadi satu kode status
kode_status_3015 = bit_member_3015 | bit_belanja_3015 | bit_barang_3015 | bit_promo_3015

# AND (&) -> memeriksa apakah bit tertentu aktif pada kode status
cek_member_bit_3015 = kode_status_3015 & 0b0001   # cek bit member
cek_promo_bit_3015 = kode_status_3015 & 0b1000    # cek bit promo tersedia

# XOR (^) -> membandingkan perbedaan status terhadap kode referensi
kode_referensi_3015 = 0b1011
perbedaan_status_3015 = kode_status_3015 ^ kode_referensi_3015

# Bonus: SHIFT (<<) -> menggeser bit kode status ke kiri 1 posisi
kode_shift_3015 = kode_status_3015 << 1

# Hak akses pelanggan ditentukan dari hasil pemeriksaan bit
member_access_3015 = bool(cek_member_bit_3015)
promo_access_3015 = bool(cek_promo_bit_3015)
free_shipping_access_3015 = bool(kode_status_3015 & 0b1000) and (kode_promo_3015 == "GRATISONGKIR")

# =====================================================================
# 9. OUTPUT / TAMPILAN HASIL PROGRAM
# =====================================================================
print()
print("=== DATA TRANSAKSI ===")
print(f"Nama Pelanggan       : {nama_3015}")
print(f"Status Pelanggan     : {status_3015}")
print(f"Total Belanja        : Rp{total_belanja_3015:.0f}")
print(f"Jumlah Barang        : {jumlah_barang_3015}")
print(f"Kode Promo           : {kode_promo_3015}")

print()
print("=== HASIL VALIDASI ===")
print(f"Belanja >= Rp200000        : {syarat_belanja_3015}")
print(f"Jumlah Barang >= 3         : {syarat_barang_3015}")
print(f"Status Member              : {is_member_3015}")
print(f"Bukan Member (not)         : {bukan_member_3015}")
print(f"Kode Promo Tersedia        : {promo_tersedia_3015}")
print(f"Kode Promo Tidak Tersedia  : {promo_tidak_tersedia_3015}")
print(f"Mendapatkan Diskon Member  : {diskon_member_3015}")
print(f"Mendapatkan Promo Tambahan : {promo_tambahan_3015}")
print(f"Mendapatkan Promo          : {mendapatkan_promo_3015}")

print()
print("=== HASIL PERHITUNGAN ===")
print(f"Diskon                     : Rp{diskon_rupiah_3015:.0f}")
print(f"Total Pembayaran           : Rp{total_pembayaran_3015:.0f}")
print(f"Rata-rata Harga Barang     : Rp{rata_rata_harga_3015:.2f}")
print(f"Sisa Bagi Barang % 3       : {sisa_bagi_3015}")
print(f"Poin Pelanggan (final)     : {poin_3015}")

print()
print("=== HAK AKSES PELANGGAN ===")
print(f"Kode Status (biner)        : {format(kode_status_3015, '04b')}")
print(f"Kode Status (desimal)      : {kode_status_3015}")
print(f"Member Access              : {member_access_3015}")
print(f"Promo Access               : {promo_access_3015}")
print(f"Free Shipping Access       : {free_shipping_access_3015}")

print()
print("=== OPERASI BITWISE ===")
print("Kode Status Transaksi (0001|0010|0100|1000 digabung dengan OR)")
print(f"Kode Biner   : {format(kode_status_3015, '04b')}")
print(f"Kode Desimal : {kode_status_3015}")

print()
print("Cek Member (AND dengan 0001)")
print(f"{format(kode_status_3015, '04b')} & 0001")
print(f"Hasil Biner   : {format(cek_member_bit_3015, '04b')}")
print(f"Hasil Desimal : {cek_member_bit_3015}")

print()
print("Cek Promo (AND dengan 1000)")
print(f"{format(kode_status_3015, '04b')} & 1000")
print(f"Hasil Biner   : {format(cek_promo_bit_3015, '04b')}")
print(f"Hasil Desimal : {cek_promo_bit_3015}")

print()
print("Perbandingan Status (XOR dengan kode referensi 1011)")
print(f"Kode Transaksi : {format(kode_status_3015, '04b')}")
print(f"Kode Referensi : {format(kode_referensi_3015, '04b')}")
print(f"{format(kode_status_3015, '04b')} ^ {format(kode_referensi_3015, '04b')}")
print(f"Hasil Biner   : {format(perbedaan_status_3015, '04b')}")
print(f"Hasil Desimal : {perbedaan_status_3015}")

print()
print("Shift Kiri (<<)")
print(f"{format(kode_status_3015, '04b')} << 1")
print(f"Hasil Biner   : {format(kode_shift_3015, '05b')}")
print(f"Hasil Desimal : {kode_shift_3015}")

print()
print("=== OPERATOR IDENTITAS (is / is not) ===")
print(f"list_a_3015 is list_b_3015     : {cek_is_3015}  (objek berbeda meski isi sama)")
print(f"list_a_3015 is not list_b_3015 : {cek_is_not_3015}")
print(f"list_a_3015 == list_b_3015     : {cek_nilai_sama_3015}  (== membandingkan isi/nilai)")

print()
print("=== SELESAI ===")

```

bit_member_3015 = 0b0001 if is_member_3015 else 0
bit_belanja_3015 = 0b0010 if syarat_belanja_3015 else 0
bit_barang_3015 = 0b0100 if syarat_barang_3015 else 0
bit_promo_3015 = 0b1000 if promo_tersedia_3015 else 0

kode_status_3015 = bit_member_3015 | bit_belanja_3015 | bit_barang_3015 | bit_promo_3015
cek_member_bit_3015 = kode_status_3015 & 0b0001
cek_promo_bit_3015 = kode_status_3015 & 0b1000
kode_referensi_3015 = 0b1011
perbedaan_status_3015 = kode_status_3015 ^ kode_referensi_3015
kode_shift_3015 = kode_status_3015 << 1

a. Setiap kondisi pelanggan direpresentasikan sebagai bit sesuai posisi: 0001=member, 0010=belanja cukup, 0100=barang cukup, 1000=promo tersedia.

b. | (OR bitwise) menggabungkan keempat bit menjadi satu kode status desimal. Jika semua kondisi True, hasilnya 1111 = 15.

c. & (AND bitwise) memeriksa apakah bit tertentu "menyala" pada kode status — misalnya kode_status & 0001 hanya bernilai bukan-nol jika bit member aktif.

4. ^ (XOR bitwise) membandingkan kode status transaksi dengan kode referensi (1011); hasilnya menunjukkan bit yang berbeda di antara keduanya (bit yang sama → 0, bit yang berbeda → 1).

5. << (shift kiri) menggeser seluruh bit kode status satu posisi ke kiri (setara mengalikan dengan 2), ditambahkan sebagai bonus demonstrasi operator bitwise lain.
member_access_3015, promo_access_3015, dan free_shipping_access_3015 mengonversi hasil & menjadi boolean (bool(...)) untuk menentukan hak akses pelanggan.

10. Bagian Output (print)

Seluruh bagian akhir program hanya menampilkan hasil dari variabel-variabel yang sudah dihitung di atas, dikelompokkan menjadi: Data Transaksi, Hasil Validasi, Hasil Perhitungan, Hak Akses Pelanggan, Operasi Bitwise (langkah demi langkah: OR → AND member → AND promo → XOR → shift), dan terakhir Operator Identitas. Fungsi format(nilai, '04b') mengubah angka desimal menjadi representasi biner 4 digit (misal 15 → 1111) agar sesuai format tampilan pada contoh soal.
